# Exploratory Data Analysis (EDA)

In this notebook we explore the NYC Airbnb dataset to understand its structure, identify missing values, detect outliers, and apply basic fixes before moving to the cleaning step of the pipeline.

```python
import wandb
import pandas as pd

run = wandb.init(project="nyc_airbnb", group="eda", save_code=True)
local_path = wandb.use_artifact("sample.csv:latest").file()
df = pd.read_csv(local_path)
```

## Profile Report

We use `ydata-profiling` to generate a comprehensive overview of the dataset, including statistics, distributions, correlations, and missing value analysis.

```python
import ydata_profiling

profile = ydata_profiling.ProfileReport(df)
profile.to_widgets()
```

## Observations

From the profiling report we can see:

- There are **missing values** in several columns (e.g., `name`, `host_name`, `last_review`, `reviews_per_month`).
- The `last_review` column is stored as a **string** instead of a datetime type.
- The `price` column contains **outliers**: some values are zero or extremely high.

After discussion with stakeholders, we will restrict nightly prices to between **$10 and $350**.

## Apply Basic Fixes

We remove price outliers and convert `last_review` to datetime.

```python
# Drop outliers
min_price = 10
max_price = 350
idx = df['price'].between(min_price, max_price)
df = df[idx].copy()

# Convert last_review to datetime
df['last_review'] = pd.to_datetime(df['last_review'])
```

## Verify Fixes

We check that prices are within the defined range and that `last_review` is now of datetime type.

```python
df.info()
```

```python
df['price'].describe()
```

Prices are now within the $10-$350 range, and `last_review` is of `datetime64` type.
